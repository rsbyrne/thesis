```python
from matplotlib import pyplot as plt
```

```python
import numpy as np
```

```python
def func(eta_ref, eta_delta, T):
    return eta_ref + eta_delta ** (1-T)

def approximation(a, corrected_eta_delta, T):
    return a * corrected_eta_delta ** (1 - T)

def correction(r, eta_ref, eta_delta):
    # a = ((1 + np.sqrt(eta_delta)) / np.sqrt(1 + eta_delta))**2
    a = eta_ref + 1
    corrected_eta_delta = ((1 + eta_delta**r) / a)**(1/r)
    return a, corrected_eta_delta

T_vals = np.linspace(0., 1, 100)
r = 1.
eta_ref = 1
eta_delta = 3e4
a, corrected_eta_delta = correction(r, eta_ref, eta_delta)

nu_vals = func(eta_ref, eta_delta, T_vals)
corrected_nu_vals = approximation(a, corrected_eta_delta, T_vals)

plt.plot(T_vals, nu_vals)
plt.plot(
    T_vals, corrected_nu_vals,
    linestyle='--'
    )
```
