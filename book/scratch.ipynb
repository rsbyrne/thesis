### Noums and the problem of ordering

We just introduced the idea of using an 'eight-space' to store an eight-bit integer. This presupposes that the dimensions are ordered, but in the Pleroma as we have defined it, not only are the dimensions not naturally ordered: they are not even naturally identifiable.

It might seem like the Pleroma, being a symmetric space, would be incapable of supporting higher structures - after all, structure implies disparity. In reality, the symmetry of the Pleroma breaks down as soon as we introduce 'content' to it.

First, let us introduce a little notation. 'Points' in the Pleroma shall now be called 'Noums'. Noums are technically infinite multisets, but if we omit all Void values (i.e. if we only count the 'active' dimensions) we can represent them much more parsimoniously:

$$
\lgroup 0.1, 0.3 \rgroup := \{0.1, 0.3, \mathrm{Void}...\}
$$

Now, take a pair of Noums, $A$ and $B$:

$$
A = \lgroup 0.1, 0.3 \rgroup
\\
B = \lgroup 0.3, 0.1 \rgroup
$$

Since the Pleroma is unordered, it is tempting to declare that these two points must be the same. However, we cannot assume this because the $0.1$ value on $B$ might be in a different dimension to the $0.1$ value on $A$. In this case, we can only say that $A$ and $B$ are the same *up to* a permutation: i.e. there is a way of allocating dimensions to $A$ and $B$ such that the two points exactly coincide. That is a *kind* of equivalence for sure, but not a strong one. We might say that Noums in such a state are 'Free Noums' because they are not attached to any particular dimension. It is apparent that Free Noums are in a sense incomparable unless we prescribe some policy that brings them into a mutual framing.

Consider $A$ and $B$. We *could* allocate four dimensions - two for each Noum - in which case one can only be reached by along a four-dimensional vector. Alternatively, we *could* allocate three dimensions - indeed, in a number of different ways - with the consequence that $A$ and $B$ would share one dimension while each having another dimension to themselves. We could even allocate these dimensions such that $A$ and $B$ express like values in the shared dimension. That would certainly be more parsimonious, but the most parsimonious allocation of all would allocate two dimensions for both Noums to share, and furthermore, with like values expressed on the same dimension. If we do this, we find that $A$ and $B$ exactly coincide: under *this dimensionalisation*, they are the same. We say that the 'free' Noums are now 'bound' (or rather, we have created a new pair of 'Bound Noums' that correspond to the Free Noums under this particular dimensionalisation).

Once we are agreed on a common policy for allocating dimensions to arbitrary sets, we can do the same thing in reverse: instead of defining dimensions to bring points together, we can *define points that bring dimensions together*. This insight gives us the foundations of a general method for structuring the Pleroma into something suitable for a given use case without contaminating it with any particular use case.

The choice of 'policy' we make here has profound implications: it must be considered carefully.

```python
import re
import os

def convert_citations(input_file, output_file=None):

    if output_file is None:
        output_file = input_file
    
    with open(input_file, 'r', encoding='utf-8') as f:
        content = f.read()

    # Replace {cite}`...` with [@...; @...] for multiple citations
    def replace_cite(match):
        citations = match.group(1).split(',')
        return '[' + '; '.join(f'@{cite.strip()}' for cite in citations) + ']'

    converted = re.sub(r'\{cite\}`([^`]+)`', replace_cite, content)

    with open(output_file, 'w', encoding='utf-8') as f:
        f.write(converted)
```
