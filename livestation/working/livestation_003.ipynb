````python
from matplotlib import pyplot as plt
import numpy as np
````

````python
r = 20
tau_ref_a = 2
tau_delta_a = 3

def curve_a(d):
    return r * (tau_ref_a + tau_delta_a ** d)

def curve_b(d):
    tau_ref_b = r * tau_ref_a + r - 1
    tau_delta_b = r * tau_delta_a - r + 1
    return tau_ref_b + tau_delta_b ** d

d_vals = np.linspace(-0.5, 1.3, 100)
nu_a_vals = curve_a(d_vals)
nu_b_vals = curve_b(d_vals)

plt.plot(d_vals, nu_a_vals)
plt.plot(d_vals, nu_b_vals)
````

````python
alpha = 1e4
tau_ref_star = 2
tau_delta_star = 3

def curve(d):
    return alpha * (tau_ref_star + tau_delta_star**d)

def curve_star(d):
    tau_ref = tau_ref_star * alpha
    if -0.01 < d < 0.01:
        return tau_ref + 1
    tau_delta = tau_delta_star * alpha ** (1/d)
    return tau_ref + tau_delta ** d

d_vals = np.linspace(-0.5, 1.3, 100)
nu_vals = curve(d_vals)
nu_star_vals = tuple(map(curve_star, tuple(d_vals)))

plt.plot(d_vals, nu_vals)
plt.plot(d_vals, nu_star_vals, linestyle='--')
````
